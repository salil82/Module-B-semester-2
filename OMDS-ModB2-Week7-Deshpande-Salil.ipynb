```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import numpy as np
import matplotlib.pyplot as plt

# Data: months and the two series (Received = gray, Processed = blue)
months = np.arange(1, 13)
month_labels = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# Values matching the example (adjust if needed)
received = np.array([160, 180, 240, 150, 180, 160, 125, 202, 160, 139, 149, 177])
processed = np.array([155, 180, 235, 150, 175, 155, 123, 156, 126, 104, 124, 140])

# Create figure
plt.figure(figsize=(11, 4.5))
ax = plt.gca()

# Plot lines connecting the actual month points (no smoothing)
ax.plot(months, received, color="#7f8185", linewidth=4.0, solid_capstyle='round', zorder=1)
ax.plot(months, processed, color="#1f2f6f", linewidth=4.0, solid_capstyle='round', zorder=2)

# Plot markers at actual month points
ax.plot(months, received, 'o', color="#7f8185", markersize=8, zorder=3)
ax.plot(months, processed, 'o', color="#1f2f6f", markersize=8, zorder=4)

# Annotate numeric values for Aug–Dec near the points
annot_months = np.array([8, 9, 10, 11, 12])
recv_annot = received[annot_months - 1]
proc_annot = processed[annot_months - 1]

# Place annotations slightly offset from points
for m, val in zip(annot_months, recv_annot):
    ax.text(m + 0.12, val + 6, str(int(val)), color="#7f8185", fontsize=12, va='center')

for m, val in zip(annot_months, proc_annot):
    ax.text(m - 0.38, val - 10, str(int(val)), color="#1f2f6f", fontsize=11, va='center')

# Axis limits, ticks and labels
ax.set_xlim(0.6, 12.4)
ax.set_ylim(0, 300)
ax.set_xticks(months)
ax.set_xticklabels(month_labels, fontsize=12)
ax.set_yticks(np.arange(0, 301, 50))
ax.tick_params(axis='y', labelsize=10)

# Style spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(2.0)
ax.spines['left'].set_color("#6b6b6b")
ax.spines['bottom'].set_linewidth(1.0)
ax.spines['bottom'].set_color("#6b6b6b")

# Place the big labels "Received" and "Processed" on the right
# Use the last actual month values for vertical placement
ax.text(12.6, received[-1], "Received", color="#7f8185",
        fontsize=28, va='center', ha='left')
ax.text(12.6, processed[-1] - 18, "Processed", color="#1f2f6f",
        fontsize=28, va='center', ha='left')

# Make room on the right for labels
plt.subplots_adjust(right=0.78)

plt.tight_layout()
plt.show()
```
